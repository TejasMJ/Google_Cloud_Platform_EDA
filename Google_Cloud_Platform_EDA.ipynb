#### Based on the GCP usage and cost report, the meaning of each column is as follows:

- **Resource ID**: Unique identifier for the GCP resource being used. (Text)  
- **Service Name**: Name of the Google Cloud service (e.g., Compute Engine, Cloud Storage) associated with the usage. (Text)  
- **Usage Quantity**: Amount of the resource consumed during the billing period (e.g., number of VM hours, GB of storage). (Numeric)  
- **Usage Unit**: Unit in which the resource usage is measured (e.g., hours, GB, requests). (Text)  
- **Region/Zone**: Geographic region or zone where the resource is located (e.g., `us-central1`, `asia-south1`). (Text)  
- **CPU Utilization (%)**: Average CPU usage percentage for the resource over the billing period. (Numeric)  
- **Memory Utilization (%)**: Average memory usage percentage for the resource over the billing period. (Numeric)  
- **Network Inbound Data (Bytes)**: Amount of incoming network traffic in bytes. (Numeric)  
- **Network Outbound Data (Bytes)**: Amount of outgoing network traffic in bytes. (Numeric)  
- **Usage Start Date**: Start date of the usage period. (Date)  
- **Usage End Date**: End date of the usage period. (Date)  
- **Cost per Quantity ($)**: Price of a single unit of the resource. (Numeric)  
- **Unrounded Cost ($)**: Raw cost calculated before rounding. (Numeric)  
- **Rounded Cost ($)**: Cost rounded to the nearest cent. (Numeric)  
- **Total Cost (INR)**: Total cost of the resource usage converted to Indian Rupees. (Numeric)

#### Observations on Data Types

From the dataset inspection, most columns are stored in appropriate data types for analysis.  
However, two columns require type conversion for proper analysis.

- **Usage Start Date** and **Usage End Date** are currently stored as `str` but should be converted to `datetime` for proper time-based analysis.

All other columns appear correctly typed:

- Identifier and categorical columns such as Resource ID, Service Name, Usage Unit, and Region/Zone are stored as `str`, which is appropriate.

- Numerical feature columns such as Usage Quantity, CPU Utilization (%), Memory Utilization (%), Network Outbound Data (Bytes), Cost per Quantity, and Unrounded Cost are stored as `float64`, which is correct.

- Discrete numerical columns such as **Network Inbound Data (Bytes)**, **Rounded Cost ($)**, and **Total Cost (INR)** are stored as `int64`, which is also appropriate.

## Data Cleaning

```python
# Convert columns to datetime
df['Usage Start Date'] = pd.to_datetime(df['Usage Start Date'], format='%d-%m-%Y %H:%M', errors='coerce')
df['Usage End Date'] = pd.to_datetime(df['Usage End Date'], format='%d-%m-%Y %H:%M', errors='coerce')

# Convert numeric columns
numeric_cols = [
    'Usage Quantity', 'CPU Utilization (%)', 'Memory Utilization (%)',
    'Network Inbound Data (Bytes)', 'Network Outbound Data (Bytes)',
    'Cost per Quantity ($)', 'Unrounded Cost ($)', 'Rounded Cost ($)', 'Total Cost (INR)'
]
df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

# Convert categorical columns
categorical_cols = ['Resource ID', 'Service Name', 'Usage Unit', 'Region/Zone']
df[categorical_cols] = df[categorical_cols].astype('category')
```

```python
df.dtypes
```

## Feature Engineering

#### To ensure the cost analysis is accessible and comparable for a global audience, we are converting the Total Cost from Indian Rupees (INR) to US Dollars (USD) using the applicable exchange rate.

```python
# Example exchange rate
exchange_rate = 83  # 1 USD = 83 INR

# Convert INR to USD
df["Total Cost (USD)"] = df["Total Cost (INR)"] / exchange_rate

# Optional: round to 2 decimal places
df["Total Cost (USD)"] = df["Total Cost (USD)"].round(2)

# Check result
df[["Total Cost (INR)", "Total Cost (USD)"]].head()
```

## Time for Analysis

#### Univaraite Analysis

```python
# -------- CATEGORICAL VARIABLES --------

# 1. Top Services
top_services = df['Service Name'].value_counts().head(20).reset_index()
top_services.columns = ['Service Name', 'Count']
fig = px.bar(
    top_services,
    x='Count',
    y='Service Name',
    orientation='h',
    title='A Few Core Services Drive the Majority of Cloud Consumption',
    text='Count'
)
fig.update_traces(textposition='outside')
fig.update_layout(yaxis=dict(categoryorder='total ascending'))
fig.show()

# 2. Region/Zone
region_counts = df['Region/Zone'].value_counts().reset_index()
region_counts.columns = ['Region/Zone', 'Count']
fig = px.bar(
    region_counts,
    x='Count',
    y='Region/Zone',
    orientation='h',
    title='Workloads Concentrate in Strategic Geographic Regions',
    text='Count'
)
fig.update_traces(textposition='outside')
fig.update_layout(yaxis=dict(categoryorder='total ascending'))
fig.show()

# 3. Usage Unit
usage_unit_counts = df['Usage Unit'].value_counts().reset_index()
usage_unit_counts.columns = ['Usage Unit', 'Count']
fig = px.bar(
    usage_unit_counts,
    x='Count',
    y='Usage Unit',
    orientation='h',
    title='Resource Measurement Units Reflect Diverse Workload Types',
    text='Count'
)
fig.update_traces(textposition='outside')
fig.show()

# 4. Top Cost-Causing Resources
top_resources = df.groupby('Resource ID')['Total Cost (USD)'].sum().sort_values(ascending=False).head(20).reset_index()
fig = px.bar(
    top_resources,
    x='Total Cost (USD)',
    y='Resource ID',
    orientation='h',
    title='A Small Set of Resources Contribute Disproportionately to Spending',
    text='Total Cost (USD)'
)
fig.update_traces(textposition='outside')
fig.update_layout(yaxis=dict(categoryorder='total ascending'))
fig.show()
```

```python
# -------- NUMERIC VARIABLES --------

# 1. Usage Quantity
fig = px.histogram(
    df, x='Usage Quantity',
    title='Cloud Consumption Is Skewed Toward High-Volume Workloads'
)
fig.show()

# 2. CPU Utilization (%)
fig = px.histogram(
    df, x='CPU Utilization (%)',
    title='Many Instances Operate Below Optimal CPU Capacity'
)
fig.show()

# 3. Memory Utilization (%)
fig = px.histogram(
    df, x='Memory Utilization (%)',
    title='Memory Usage Patterns Reveal Over-Provisioning Opportunities'
)
fig.show()

# 4. Network Inbound Data (Bytes)
fig = px.histogram(
    df, x='Network Inbound Data (Bytes)',
    title='Inbound Traffic Varies Widely Across Services'
)
fig.show()

# 5. Network Outbound Data (Bytes)
fig = px.histogram(
    df, x='Network Outbound Data (Bytes)',
    title='Outbound Traffic May Be a Hidden Cost Driver'
)
fig.show()

# 6. Cost per Quantity ($)
fig = px.histogram(
    df, x='Cost per Quantity ($)',
    title='Pricing Varies Significantly Across Service Categories'
)
fig.show()

# 7. Unrounded Cost ($)
fig = px.histogram(
    df, x='Unrounded Cost ($)',
    title='Raw Cost Distribution Highlights Heavy-Tail Spending'
)
fig.show()

# 8. Rounded Cost ($)
fig = px.histogram(
    df, x='Rounded Cost ($)',
    title='Billing Rounding Has Minimal but Noticeable Financial Impact'
)
fig.show()

# 9. Total Cost (USD)
fig = px.histogram(
    df, x='Total Cost (USD)',
    title='A Small Percentage of Resources Account for Most of the Spend'
)
fig.show()
```

```python
# -------- DATE VARIABLES --------

# Derived column: Usage Duration
df['Usage Duration'] = (df['Usage End Date'] - df['Usage Start Date']).dt.total_seconds()/3600  # hours

# Usage Start Date
fig = px.line(
    df.groupby('Usage Start Date')['Total Cost (USD)'].sum().reset_index(),
    x='Usage Start Date', y='Total Cost (USD)',
    title='Cloud Usage Shows Periodic Spikes and Demand Cycles'
)
fig.show()

# Usage End Date
fig = px.histogram(
    df, x='Usage End Date',
    title='Billing Period Length Influences Cost Concentration'
)
fig.show()

# Usage Duration
fig = px.histogram(
    df, x='Usage Duration',
    title='Long-Running Resources Drive Sustained Cost Exposure'
)
fig.show()
```

#### Bivariate Analysis

```python
# -------- Numeric vs Numeric --------

# CPU vs Memory Utilization
fig = px.scatter(
    df, x='CPU Utilization (%)', y='Memory Utilization (%)',
    title='Compute and Memory Usage Are Not Always Aligned'
)
fig.show()

# Usage Quantity vs Total Cost
fig = px.scatter(
    df, x='Usage Quantity', y='Total Cost (USD)',
    title='Higher Consumption Directly Drives Cost Escalation'
)
fig.show()

# Network Outbound vs Total Cost
fig = px.scatter(
    df, x='Network Outbound Data (Bytes)', y='Total Cost (USD)',
    title='Data Egress Significantly Impacts Overall Spending'
)
fig.show()

# CPU Utilization vs Total Cost
fig = px.scatter(
    df, x='CPU Utilization (%)', y='Total Cost (USD)',
    title='Underutilized Compute Still Generates High Costs'
)
fig.show()

# Usage Duration vs Total Cost
fig = px.scatter(
    df, x='Usage Duration', y='Total Cost (USD)',
    title='Longer Resource Lifetimes Correlate With Higher Spending'
)
fig.show()
```

```python
# -------- Numeric vs Categorical --------

# Average Total Cost by Service Name
avg_cost_service = df.groupby('Service Name')['Total Cost (USD)'].mean().sort_values(ascending=False).reset_index()
fig = px.bar(
    avg_cost_service, x='Total Cost (USD)', y='Service Name', orientation='h',
    title='Compute Services Dominate Overall Cloud Expenditure', text='Total Cost (USD)'
)
fig.update_traces(textposition='outside')
fig.update_layout(yaxis=dict(categoryorder='total ascending'))
fig.show()

# Average CPU Utilization by Service Name
avg_cpu_service = df.groupby('Service Name')['CPU Utilization (%)'].mean().sort_values().reset_index()
fig = px.bar(
    avg_cpu_service, x='CPU Utilization (%)', y='Service Name', orientation='h',
    title='Some Services Show Consistently Low Resource Efficiency', text='CPU Utilization (%)'
)
fig.update_traces(textposition='outside')
fig.update_layout(yaxis=dict(categoryorder='total ascending'))
fig.show()

# Average Total Cost by Region/Zone
avg_cost_region = df.groupby('Region/Zone')['Total Cost (USD)'].mean().sort_values(ascending=False).reset_index()
fig = px.bar(
    avg_cost_region, x='Total Cost (USD)', y='Region/Zone', orientation='h',
    title='Regional Deployment Strategy Impacts Cost Structure', text='Total Cost (USD)'
)
fig.update_traces(textposition='outside')
fig.update_layout(yaxis=dict(categoryorder='total ascending'))
fig.show()
```

```python
# -------- CATEGORICAL VS CATEGORICAL --------

# Service Name vs Region/Zone
service_region = df.groupby(['Service Name','Region/Zone']).size().reset_index(name='Count')
fig = px.bar(
    service_region,
    x='Service Name',
    y='Count',
    color='Region/Zone',
    title='Specific Services Are Concentrated in Select Regions',
    text='Count'
)
fig.update_traces(textposition='outside')
fig.update_layout(barmode='stack', xaxis_tickangle=45)
fig.show()

# Service Name vs Usage Unit
service_unit = df.groupby(['Service Name','Usage Unit']).size().reset_index(name='Count')
fig = px.bar(
    service_unit,
    x='Service Name',
    y='Count',
    color='Usage Unit',
    title='Different Services Reflect Distinct Consumption Models',
    text='Count'
)
fig.update_traces(textposition='outside')
fig.update_layout(barmode='stack', xaxis_tickangle=45)
fig.show()

# Region vs High-Cost Resources (Top 10%)
threshold = df['Total Cost (USD)'].quantile(0.9)
high_cost_resources = df[df['Total Cost (USD)'] >= threshold]
region_highcost = high_cost_resources.groupby('Region/Zone')['Resource ID'].nunique().reset_index()
region_highcost.columns = ['Region/Zone', 'High-Cost Resources']
fig = px.bar(
    region_highcost,
    x='Region/Zone',
    y='High-Cost Resources',
    title='High-Spend Resources Cluster in Key Strategic Regions',
    text='High-Cost Resources'
)
fig.update_traces(textposition='outside')
fig.show()
```

#### Multivaraite Analysis

```python
# -------------------------
# MULTIVARIATE ANALYSIS
# -------------------------

# Total Cost × Service Name × Region
cost_service_region = df.groupby(['Service Name','Region/Zone'])['Total Cost (USD)'].sum().reset_index()
fig = px.density_heatmap(
    cost_service_region, x='Service Name', y='Region/Zone', z='Total Cost (USD)',
    title='Spending Concentrates Around Specific Services in Key Regions'
)
fig.show()

# CPU × Memory × Service Name
fig = px.scatter(
    df, x='CPU Utilization (%)', y='Memory Utilization (%)', size='Total Cost (USD)',
    color='Service Name', hover_data=['Resource ID'],
    title='Efficiency Profiles Differ Across Service Categories'
)
fig.show()

# Total Cost × Usage Quantity × Service Name
fig = px.scatter(
    df, x='Usage Quantity', y='Total Cost (USD)', size='Total Cost (USD)',
    color='Service Name', hover_data=['Resource ID'],
    title='High-Volume Services Are Primary Revenue Drivers'
)
fig.show()

# Network Outbound × Region × Total Cost
fig = px.scatter(
    df, x='Region/Zone', y='Network Outbound Data (Bytes)', size='Total Cost (USD)',
    color='Region/Zone', hover_data=['Service Name'],
    title='Data Egress Patterns Explain Regional Cost Variations'
)
fig.show()

# Usage Duration × CPU Utilization × Total Cost
fig = px.scatter(
    df, x='Usage Duration', y='CPU Utilization (%)', size='Total Cost (USD)',
    color='Service Name', hover_data=['Resource ID'],
    title='Long-Running, Underutilized Resources Represent Optimization Opportunities'
)
fig.show()
```
